**Ejemplo de invocación a GPT-4o-mini**

```python
from openai import AzureOpenAI

endpoint = "https://guarania-maas.cognitiveservices.azure.com/"
model_name = "gpt-4o-mini"
deployment = "gpt-4o-mini"

subscription_key = userdata.get('AZURE_API_KEY')
api_version = "2024-12-01-preview"

client = AzureOpenAI(
    api_version=api_version,
    azure_endpoint=endpoint,
    api_key=subscription_key,
)

response = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": "Eres un asistente.",
        },
        {
            "role": "user",
            "content": "Escribe un poema corto sobre la inteligencia artificial.",
        }
    ],
    max_tokens=4096,
    temperature=1.0,
    top_p=1.0,
    model=deployment
)

print(response.choices[0].message.content)
```

**Grok-4.1-Fast-Non Reasoning**

```python
from openai import OpenAI

endpoint = "https://guarania-maas.services.ai.azure.com/openai/v1/"
model_name = "grok-4-1-fast-non-reasoning"
deployment_name = "grok-4-1-fast-non-reasoning"

api_key = userdata.get('AZURE_API_KEY')

client = OpenAI(
    base_url=f"{endpoint}",
    api_key=api_key
)

completion = client.chat.completions.create(
    model=deployment_name,
    messages=[
        {
            "role": "user",
            "content": "Escribe un poema corto sobre la inteligencia artificial.",
        }
    ],
)

print(completion.choices[0].message)
```
